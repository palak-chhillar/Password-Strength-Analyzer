```python
def check_password(password):
    checks = {
        "Length (8+ characters)": len(password) >= 8,
        "Uppercase letter": bool(re.search(r"[A-Z]", password)),
        "Lowercase letter": bool(re.search(r"[a-z]", password)),
        "Number": bool(re.search(r"[0-9]", password)),
        "Special character": bool(re.search(r"[^A-Za-z0-9]", password))
    }

    score = sum(checks.values())

    return checks, score
```

```python
def get_strength(score):
    if score <= 2:
        return "Weak"
    elif score == 3:
        return "Moderate"
    elif score == 4:
        return "Strong"
    else:
        return "Very Strong"
```

```python
password = input("Enter a test password: ")

checks, score = check_password(password)
strength = get_strength(score)
is_common = check_common_password(password)
has_repeats = has_repeated_characters(password)
has_sequence = has_sequential_characters(password)

print("\n🔐 Password Analysis")
print("-" * 30)

for criterion, passed in checks.items():
    print(f"{'✅' if passed else '❌'} {criterion}")

print(f"\nStrength: {strength}")
print(f"Score: {score}/5")

if is_common:
    print("⚠️ Warning: This is a commonly used password!")
else:
    print("✅ Password is not in the common password list.")

if has_repeats:
    print("⚠️ Warning: Password contains many repeated characters!")
else:
    print("✅ No excessive character repetition detected.")

if has_sequence:
    print("⚠️ Warning: Password contains a sequential pattern!")
else:
    print("✅ No obvious sequential pattern detected.")
```

```python
common_passwords = [
    "password",
    "123456",
    "12345678",
    "qwerty",
    "admin",
    "welcome",
    "password123",
    "abc123",
    "letmein",
    "iloveyou"
]
```

```python
is_common = password.lower() in common_passwords

if is_common:
    print("⚠️ Warning: This is a commonly used password!")
else:
    print("✅ This password is not in the common password list.")
```

```python
def check_common_password(password):
    return password.lower() in common_passwords
```

```python
def has_repeated_characters(password):
    return len(set(password)) < len(password) * 0.5
```

```python
def has_sequential_characters(password):
    sequences = [
        "0123456789",
        "abcdefghijklmnopqrstuvwxyz",
        "qwertyuiop"
    ]

    password = password.lower()

    for sequence in sequences:
        for i in range(len(sequence) - 2):
            if sequence[i:i+3] in password:
                return True

    return False
```
